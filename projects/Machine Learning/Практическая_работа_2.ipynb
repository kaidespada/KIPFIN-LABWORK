### Задание 5: Описание набора данных

- **Product_Category_2** - вторая категория товара (может быть пустой)
- **Product_Category_3** - третья категория товара (может быть пустой)
- **Purchase** - сумма покупки

**Основные характеристики:**
- Набор данных содержит информацию о множественных покупках разными пользователями
- Присутствуют пропущенные значения в столбцах Product_Category_2 и Product_Category_3
- Данные представляют собой транзакционную историю покупок

### Задание 6: Сколько всего возрастных категорий?

```python
age_categories = df['Age'].unique()
print("Возрастные категории:")
print(age_categories)
print(f"\nВсего возрастных категорий: {len(age_categories)}")
```

### Задание 7: Сколько строк с мужчинами из категории города А?

```python
# Фильтрация: мужчины (Gender == 'M') и город категории А (City_Category == 'A')
men_city_a = df[(df['Gender'] == 'M') & (df['City_Category'] == 'A')]
count_men_city_a = len(men_city_a)

print(f"Количество строк с мужчинами из категории города А: {count_men_city_a}")
```

### Задание 8: Сколько женщин от 46 до 50, потративших больше 20000?

```python
# Фильтрация: женщины (Gender == 'F'), возраст 46-50 (Age == '46-50'), покупка > 20000
women_46_50_high_purchase = df[
    (df['Gender'] == 'F') & 
    (df['Age'] == '46-50') & 
    (df['Purchase'] > 20000)
]
count_women = len(women_46_50_high_purchase)

print(f"Количество женщин от 46 до 50 лет, потративших больше 20000: {count_women}")
```

### Задание 9: Сколько пропущенных значений в столбце Product_Category_3?

```python
# Подсчет пропущенных значений
missing_count = df['Product_Category_3'].isna().sum()

print(f"Количество пропущенных значений в столбце Product_Category_3: {missing_count}")
print(f"Доля пропущенных значений: {missing_count / len(df) * 100:.2f}%")
```

### Задание 10: Какую долю от всех покупателей составляют вместе мужчины от 26 до 35 лет и женщины старше 36 лет?

```python
# Мужчины от 26 до 35 лет
men_26_35 = df[(df['Gender'] == 'M') & (df['Age'] == '26-35')]
count_men_26_35 = len(men_26_35)

# Женщины старше 36 лет (категории: 36-45, 46-50, 51-55, 55+)
women_over_36 = df[
    (df['Gender'] == 'F') & 
    (df['Age'].isin(['36-45', '46-50', '51-55', '55+']))
]
count_women_over_36 = len(women_over_36)

# Общее количество
total_target_group = count_men_26_35 + count_women_over_36
total_records = len(df)

# Доля от всех покупателей
proportion = total_target_group / total_records

print("=== Анализ целевой группы ===")
print(f"Мужчин от 26 до 35 лет: {count_men_26_35}")
print(f"Женщин старше 36 лет: {count_women_over_36}")
print(f"Всего в целевой группе: {total_target_group}")
print(f"Всего записей в наборе данных: {total_records}")
print(f"\nДоля целевой группы от всех покупателей: {proportion:.4f} ({proportion * 100:.2f}%)")
```
